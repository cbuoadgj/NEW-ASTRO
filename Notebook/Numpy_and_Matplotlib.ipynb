**INTRO TO PYTHON LIBRARIES**

### 1. Importing

```python
import numpy as np #importing numpy
import matplotlib.pyplot as plt



```

### 2. Convert list to np array

```python
lst=[1,2,3,4,5,6,7,8,9]
print(type(lst))
arr=np.array(lst)
print(type(arr))
```

### Create 1d array

```python
arr1=np.array([1,4,7])
print(arr1)
print(type(arr1))
```

### 3. Check data type and dimension and shape of array

```python
print(arr1.dtype) #datatype
print(arr1.ndim)  #dimension
print(arr1.shape) #shape


```

```python
# create 2D array
arr2=np.array([[1,2,4],[2,3,4.0],[9,8,7]])
print(arr2)
print(arr2.ndim)
print(arr2.dtype)
print(arr2.shape)
```

### 4. Different ways to create Arrays

```python
# 1. Arange
arange=np.arange(1,17,1)

reshaped=arange.reshape(4,4) # Reshape to 4x4

reshaped2=reshaped.astype(float) #change to float

print(reshaped2)
print(reshaped2.dtype)
print(reshaped2.ndim)
print(reshaped2.shape)
```

```python
# 2. Linspace
linspace=np.linspace(1,10,16)
lin=linspace.reshape(4,4)
print(lin)
print(lin.dtype)
print(lin.ndim)
print(lin.shape)
```

```python
# 3. Zeros
zeros=np.zeros((4,4))
print(zeros)
print(zeros.dtype)
print(zeros.ndim)
print(zeros.shape)
```

```python
# 4. Ones
ones=np.ones((3,3))
print(ones)
print(ones.dtype)
print(ones.ndim)
print(ones.shape)
```

### 5. Random numbers

```python
# 1. Rand
rand=np.random.rand() #random number between 0 and 1(not included 1)
print(rand)
if rand<0.5:
    print("Heads")
else:
    print("Tails")
```

```python
# 2. Randint
randint=np.random.randint(1,10,1)
print(randint)
```

```python
# 3. Seed
np.random.seed(42) # same type of randomness
print(np.random.randint(1,11,(2,2)))
```

### 6.Create Visualization (matplotlib)

```python
# create Quadratic Curve
x=np.linspace(-10,10,100)
y=x**3

# Show Plot
plt.figure(figsize=(8,3))
plt.grid()
plt.title("CUBIC CURVE",color='r',weight='bold',fontsize=15)
plt.xlabel("x",color='b')
plt.ylabel("y",color='b')
plt.plot(x,y)
plt.show()
```
