```python
def calc(a,b,opr):
  if opr.lower()=="add":
    return a+b
  elif opr.lower()=="sub":
    return a-b
  elif opr.lower()=="mult":
    return a*b
  elif opr.lower()=="div":
    return a/b
  else:
    return("unknow operation")

a=input("Enter value:")
a=float(a)
b=input("Enter next value:")
b=float(b)
opr=input("Enter operation:")
val=calc(a,b,opr)
print(f"Your answer is:{val}")
```
