### Method to move rocket up

```python
# by 1 unit
class rocket():

  def __init__(self): #always self as first argument
   self.x=0
   self.y=0
  def move_up(self):
   self.y+=float(input("Enter move up distance:"))

orion=rocket()
print(f"Current Position={(orion.x,orion.y)}")
orion.move_up()
print(f"moved to={(orion.x,orion.y)}")
orion.move_up()
print(f"moved to={(orion.x,orion.y)}")
```

### Change existing rocket class


```python
class rocket():

  def __init__(self,x=0,y=0): #always self as first argument
   self.x=x
   self.y=y
  def move_rocket(self,x_i=0,y_i=0):
   self.x+=x_i
   self.y+=y_i
  def get_dist(self,other_rocket):
   x_diff=self.x-other_rocket.x
   y_diff=self.y-other_rocket.y
   dist=(x_diff**2+y_diff**2)**0.5
   return round(dist,5)

rocket2=rocket(10)
rocket3=rocket()
rocket3.move_rocket(90)

print(f"{rocket2.x,rocket2.y}")
print(f"{rocket3.x,rocket3.y}")
```

### To calculate dist btw 2 rockets

```python
rocket4=rocket(-15,10)
rocket5=rocket(0,15)
print(f'Rocket4={rocket4.x,rocket4.y}')
print(f'Rocket5={rocket5.x,rocket5.y}')
dist_45=rocket4.get_dist(rocket5)
print(f'Distance btw them={dist_45}')
```
